```python
# Bibliotecas
import  os
import  yfinance                as      yf
import  pandas                  as      pd
import  numpy                   as      np
import  plotly.express          as      px
from    datetime                import  datetime, timedelta
```

```python
# Seleccionamos el par a usar

lst_name = ['SOXL', 'SOXS']
# lst_name = ['SPXL', 'SPXS']
# lst_name = ['GUSH', 'DRIP']
# lst_name = ['NUGT', 'DUST']
# lst_name = ['YINN', 'YANG']
# lst_name = ['TQQQ', 'SQQQ']
# lst_name = [ 'FAS', 'FAZ' ]
# lst_name = [ 'EDC', 'EDZ' ]
# lst_name = [ 'TNA', 'TZA' ]
# lst_name = [ 'XPP', 'FXP' ]

# lst_name = ['NVDL', 'NVD' ] # NVDA
# lst_name = ['TSLR', 'TSDD'] # TSLA
```

```python
# Obtener la hora actual
cu_time = datetime.now()                                # Current time

# Sumar 6 horas para ajustar con YF
ad_time = cu_time + timedelta(hours=6)                  # Adj time

# Darle formato para usar en YF, será nuestro End Date
en_time = ad_time
# en_date = ad_time.strftime("%Y-%m-%d_%H-%M-%S")
# en_date = ad_time.strftime("%Y-%m-%d")

# Vamos a definir nuestro Start Date
st_time = ad_time - timedelta(days=365*2)
# st_date = st_time.strftime("%Y-%m-%d_%H-%M-%S")
# st_date = st_time.strftime("%Y-%m-%d")

print( st_time, en_time )

df_data = yf.download( lst_name, start = st_time, end = en_time, auto_adjust=True, interval = "1d" ) 
df_data
```

```python
precios = df_data[ df_data.columns[[0,1]] ]
precios
```

```python
fig = px.line( np.log( precios[('Close',)] ) )
fig.add_trace( px.scatter( np.log( precios[('Close',)] ) ).data[0] )
fig.add_trace( px.scatter( np.log( precios[('Close',)] ) ).data[1] )
fig.show()
```

```python
# Guardar el gráfico como un archivo HTML
dir_gra = os.getcwd() + '/daily_pair.html'              # Filename direction
fig.write_html( dir_gra )
```
